## Rechnen mit Python

Es gibt verschiedene Möglichkeiten mit Hilfe eines Computers
arithmetische Operationen durchzuführen. Hier lernen Sie, wie Sie solche
mit Python vornehmen können.

Python stellt Ihnen die folgenden arithmetischen Operationen zur
Verfügung:

| Operation | Operationszeichen | Algebraisch | Python | Nummerisch |
| :--- | :--- | :--- | :--- | :--- |
| Addition | $+$ | $a + b$ | `4 + 5` | $4 + 5 = 9$ |
| Subtraktion | $-$ | $a - b$ | `5 - 4` | $5 - 4 = 1$ |
| Multiplikation | $\cdot$ | $a \cdot b$ | `3 * 4` | $3 \cdot 4 = 12$ |
| Potenzierung | | $a ^ b$ | `2 ** 3` | $2 ^ 3 = 8$ |
| Division | $\frac{}{}$ | $\frac{a}{b}$ | `7 / 2` | $\frac{7}{2} = 3.5$ |
| Ganzzahldivision | $\lfloor \rfloor$ | $\lfloor \frac{a}{b} \rfloor$ | `7 // 2` | $\lfloor \frac{7}{2} \rfloor = 3$ |
| Modulo | $\mod$ | $a\mod b$ | `7 % 2` | $7 \mod 2 = 1$ |

Python ist so programmiert, dass die Rangreihenfolge der verschiedenen
arithmetischen Operationen eingehalten werden.
