## Rechnen mit Python

Es gibt verschiedene Möglichkeiten mit Hilfe eines Computers
arithmetische Operationen durchzuführen. Hier lernen Sie, wie Sie solche

Python stellt Ihnen die folgenden arithmetischen Operationen zur
Verfügung:

| Operation | Operationszeichen | Algebraisch | Python | Nummerisch |

## Werte Variablen zuweisen

Die oben durchgeführten Berechnungen haben das Resultat jeweils direkt
ausgegeben. In vielen Fällen ist das Resultat einer konkreten Berechnung
jedoch lediglich ein Zwischenresultat. In diesen Fällen will man mit
diesem Zwischenresultat weiterrechnen können. Hier kommen Variablen ins
Spiel.

Die Resultate Ihrer Berechnungen werden nicht nur angezeigt, Sie werden
auch im Arbeitsspeicher abgelegt. Variablen dienen nun dazu, einen
ansprechbaren Verweis auf den Ort des abgelegten Resultates anzulegen.
Das heisst, wir können die Resultate der Berechnungen nicht nur
ausgeben, wir können sie einer Variabel zuweisen. Dazu wird in Python
das Zeichen `=` verwendet (ich habe das Zeichne in diesem Zusammenhang
bewusst nicht Gleichheitszeichen genannt weil es hier keine Gleichheit
sondern eine Zuweisung zum Ausdruck bringt). Dabei muss die Variabel
links des Zuweisungszeichens und der zuzuweisende Wert rechts davon
stehen. 

Als Variablen können grundsätzlich alle alphanumerischen Zeichen
verwendet werden. Ausgenommen sind die sogenannten '[reservierten
Wörter](https://www.w3schools.com/python/python_ref_keywords.asp)'.
Diesen reservierten Wörtern wurde durch die Entwickler von Python
bereits vorgegebenen Aufgaben zugewiesen. Sie sind in diesem Sinne als
Variablen schon gebraucht. Ausserdem dürfen Variablen nicht mit einer
Zahl beginnen. Als Konvention hat sich eingebürgert, dass
Python-Variablen in sogenannten Snake Cases geschrieben werden. Snake
Case sieht folgendermassen aus: `ich_bin_eine_python_variabel`. Dies
ermöglicht es, Variablen mit Namen zu versehen, welche eine kurze
Erklärung zu den Ihnen zugewiesenen Werten abgeben.

Für die erste Berechnung von oben kann das folgendermassen aussehen:

```python
resultat_a = 2 + 3 * 4
```

Nun wird das Resultat der Berechnung der Variabel `resultat_a`
zugewiesen und nicht mehr ausgegeben. Wir können den zugewiesenen Wert
jederzeit ausgeben, indem wir die Variabel aufrufen.

```python
resultat_a
```

Die Berechnung und der Aufruf der Variabel können auch in der gleichen
Code-Zelle stehen. Ausgegeben wird allerdings immer nur der Wert der
letzten Code-Zeile.

```python
resultat_a = 2 + 3 * 4
resultat_a
```
