```python
verlauf = leere_grafik()
for b in range(0, 256, 32):
    farbe = rgb_farbe(0, 0, b)
    quadrat = rechteck(15, 15, farbe)
    verlauf = neben(verlauf, quadrat)

zeige_grafik(verlauf)
```

```python
vertikal = leere_grafik()
for r in range(0, 256, 32):
    farbe = rgb_farbe(r, 0, 0)
    quadrat = rechteck(15, 15, farbe)
    vertikal = ueber(vertikal, quadrat)
    
zeige_grafik(vertikal)
```

```python
farbquadrat = leere_grafik()
for r in range(0, 256, 8):
    zeile = leere_grafik()
    # pro Zeile neu starten
    for b in range(0, 256, 8):
        farbe = rgb_farbe(r, 0, b)  # Rot aus der äusseren, Blau aus der inneren Schleife
        quadrat = rechteck(15, 15, farbe)
        zeile = neben(zeile, quadrat)
        
    farbquadrat = ueber(farbquadrat, zeile)

zeige_grafik(farbquadrat)
```

```python
front = rechteck(20, 20, weiss)
rahmen = rechteck(22, 22, schwarz)
front = ueberlagere(front, rahmen)
front = fixiere(oben_links, front)
zeige_grafik(front)
```

```python
# Bausteine: rechtwinklige Dreiecke (Rand schwarz, Kern weiss)
dreieck_b = dreieck(11, 11, 90, schwarz)
dreieck_b_kern = dreieck(10, 10, 90, weiss)

# Deckel: Parallelogramm aus Dreieck links, Rechteck, Dreieck rechts
deckel_l = drehe(90, dreieck_b)
deckel_l_kern = drehe(90, dreieck_b_kern)
deckel_r = drehe(-90, dreieck_b)
deckel_r_kern = drehe(-90, dreieck_b_kern)

deckel = rechteck(11, 11, schwarz)
deckel_kern = rechteck(10, 10, weiss)

deckel = neben(deckel_l, deckel)
deckel = neben(deckel, deckel_r)
deckel_kern = neben(deckel_l_kern, deckel_kern)
deckel_kern = neben(deckel_kern, deckel_r_kern)
deckel = ueberlagere(deckel_kern, deckel)
deckel = fixiere(unten_links, deckel)

# Seite: Parallelogramm aus Dreieck oben, Rechteck, Dreieck unten
seite_o = drehe(90, dreieck_b)
seite_o_kern = drehe(90, dreieck_b_kern)
seite_u = drehe(-90, dreieck_b)
seite_u_kern = drehe(-90, dreieck_b_kern)

seite = rechteck(11, 11, schwarz)
seite_kern = rechteck(10, 10, weiss)

seite = ueber(seite_o, seite)
seite = ueber(seite, seite_u)
seite_kern = ueber(seite_o_kern, seite_kern)
seite_kern = ueber(seite_kern, seite_u_kern)
seite = ueberlagere(seite_kern, seite)
seite = fixiere(oben_rechts, seite)

zeige_grafik(deckel)
```

```python
zeige_grafik(seite)
```

```python
wuerfel = kombiniere(front, deckel)
wuerfel = fixiere(oben_rechts, wuerfel)
zeige_grafik(wuerfel)
```

```python
wuerfel = kombiniere(wuerfel, seite)
zeige_grafik(wuerfel)
```
