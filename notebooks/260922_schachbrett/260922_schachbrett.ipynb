# Wiederholungen in Python II

Wiederholungen in Python können auch verschachtelt werden. Wie das
funktioniert, wird am Beispiel der Zeichnung eines Schachbretts gezeigt.
Der Einfachheit halber, werden auch für dieses Beispiel alle
deutschsprachigen Funktionen von PyTamaro importiert.

```python
from pytamaro.de import *
```

Ein Schachbrett besteht aus 64 Quadraten in zwei Farben. Inspiriert von
einem Schachbrett aus Holz, habe ich die zwei Farben `dunkelbraun` und
`elfenbein` mit der Funktion `rgb_farbe()` definiert. Anschliessend habe
ich die beiden Feldtypen dunkel und hell definiert.

```python
dunkelbraun = rgb_farbe(139,69,19)
elfenbein = rgb_farbe(255,255,240)
```

```python
feld_dunkel = rechteck(30, 30, dunkelbraun)
feld_hell = rechteck(30, 30, elfenbein)

zeige_grafik(feld_dunkel)
zeige_grafik(feld_hell)
```

Aus den Elementen `feld_dunkel` und `feld_hell` können nun die einzelnen
Zeilen des Schachbretts aufgebaut werden. Dazu werden zuerst die
nächstgrösseren Elemente bestehend aus einem dunklen und einem hellen
Feld bzw. umgekehrt zusammengestellt. Diese können anschliessend zu
Zeilen, beginnend mit einem dunklen bzw. einem hellen Feld aufgebaut werden.

```python
paar_dunkel = neben(feld_dunkel, feld_hell)
paar_hell = neben(feld_hell, feld_dunkel)

zeile_dunkel = leere_grafik()
zeile_hell = leere_grafik()

for i in range(4):
    zeile_dunkel = neben(zeile_dunkel, paar_dunkel)
    zeile_hell = neben(zeile_hell, paar_hell)
    
zeige_grafik(zeile_dunkel)
zeige_grafik(zeile_hell)
```

In einem weiteren Schirtt können die beiden Zeilen in einer Schleife zum
Schachbrett zusammengefügt werden.

```python
brett = leere_grafik()

for i in range(4):
    brett = ueber(brett, zeile_dunkel)
    brett = ueber(brett, zeile_hell)
    
zeige_grafik(brett)
```

Anstelle von zwei unabhängigen `for`-Schleifen, kann das Schachbrett
aber auch über eine verschachtelte `for`-Schleife gebaut werden.

```python
brett = leere_grafik()
```

```python
for i in range(4):
    zeile_dunkel = leere_grafik()
    zeile_hell = leere_grafik()
    paar_dunkel = neben(feld_dunkel, feld_hell)
    paar_hell = neben(feld_hell, feld_dunkel)
    for i in range(4):
        zeile_dunkel = neben(zeile_dunkel, paar_dunkel)
        zeile_hell = neben(zeile_hell, paar_hell)

    brett = ueber(brett, zeile_dunkel)
    brett = ueber(brett, zeile_hell)

zeige_grafik(brett)
```
