# Wiederholungen in Python II

```python
from pytamaro.de import *
```

```python
dunkelbraun = rgb_farbe(139,69,19)
elfenbein = rgb_farbe(255,255,240)
```

```python
feld_dunkel = rechteck(30, 30, dunkelbraun)
feld_hell = rechteck(30, 30, elfenbein)
```

```python
brett = leere_grafik()
zeile_hell = leere_grafik()
zeile_dunkel = leere_grafik()
doppelzeile = leere_grafik()
paar_dunkel = leere_grafik()
paar_hell =leere_grafik()
```

```python
for i in range(4):
    zeile_dunkel = leere_grafik()
    zeile_hell = leere_grafik()
    paar_dunkel = neben(feld_dunkel, feld_hell)
    paar_hell = neben(feld_hell, feld_dunkel)
    for i in range(4):
        zeile_dunkel = neben(zeile_dunkel, paar_dunkel)
        zeile_hell = neben(zeile_hell, paar_hell)

    brett = ueber(brett, zeile_dunkel)
    brett = ueber(brett, zeile_hell)

zeige_grafik(brett)
```
